**Operadores**

*Aritméticos:*

| Operador | Descripción      | Ejemplo  | Resultado |
| -------- | ---------------- | -------- | --------- |
| `+`      | Suma             | `5 + 3`  | `8`       |
| `-`      | Resta            | `10 - 2` | `8`       |
| `*`      | Multiplicación   | `4 * 2`  | `8`       |
| `/`      | División (float) | `9 / 2`  | `4.5`     |
| `//`     | División entera  | `9 // 2` | `4`       |
| `%`      | Módulo (residuo) | `9 % 2`  | `1`       |
| `**`     | Potencia         | `3 ** 2` | `9`       |
---

*Relacionales*
| Operador | Descripción       | Ejemplo  | Resultado |
| -------- | ----------------- | -------- | --------- |
| `==`     | Igual a           | `5 == 5` | `True`    |
| `!=`     | Diferente de      | `5 != 3` | `True`    |
| `>`      | Mayor que         | `7 > 2`  | `True`    |
| `<`      | Menor que         | `3 < 8`  | `True`    |
| `>=`     | Mayor o igual que | `5 >= 5` | `True`    |
| `<=`     | Menor o igual que | `2 <= 5` | `True`    |

---
*Lógicos*
| Operador | Descripción                       | Ejemplo               | Resultado |
| -------- | --------------------------------- | --------------------- | --------- |
| `and`    | Verdadero si ambos son verdaderos | `(5 > 3) and (2 < 4)` | `True`    |
| `or`     | Verdadero si alguno es verdadero  | `(5 < 3) or (2 < 4)`  | `True`    |
| `not`    | Invierte el valor lógico          | `not(5 > 3)`          | `False`   |

---
*De Asignación*
| Operador | Equivalente a | Ejemplo   |
| -------- | ------------- | --------- |
| `=`      | `x = 5`       | `x = 5`   |
| `+=`     | `x = x + 3`   | `x += 3`  |
| `-=`     | `x = x - 2`   | `x -= 2`  |
| `*=`     | `x = x * 2`   | `x *= 2`  |
| `/=`     | `x = x / 3`   | `x /= 3`  |
| `//=`    | `x = x // 2`  | `x //= 2` |
| `%=`     | `x = x % 2`   | `x %= 2`  |
| `**=`    | `x = x ** 3`  | `x **= 3` |

---

*Pertenencia*

| Operador | Descripción                             | Ejemplo      | Resultado |
| -------- | --------------------------------------- | ------------ | --------- |
| `is`     | Verdadero si son el mismo objeto        | `x is y`     | Depende   |
| `is not` | Verdadero si **no** son el mismo objeto | `x is not y` | Depende   |

---

*De Bit a Bit*
| Operador | Nombre  | Ejemplo  | Resultado binario |    |         |
| -------- | ------- | -------- | ----------------- | -- | ------- |
| `&`      | AND     | `5 & 3`  | `0b1`             |    |         |
| `\|`      | OR      | `5 \| 3`  | `0b111`           |    |         |
| `^`      | XOR     | `5 ^ 3`  | `0b110`           |    |         |
| `~`      | NOT     | `~5`     | `-6`              |    |         |
| `<<`     | Desplazamiento izquierda | `5 << 1` | `10`              |    |         |
| `>>`     | Desplazamiento derecha   | `5 >> 1` | `2`               |    |         |

---

````python
# Iterable personalizado

class Contador:
    def __init__(self, limite):
        self.limite = limite
        self.actual = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.actual >= self.limite:
            raise StopIteration
        valor = self.actual
        self.actual += 1
        return valor

for x in Contador(5):
    print(x)
````

````python
# con itertools

import itertools

# Definición del generador
def contador_infinito():
    count = 0
    while True:
        yield count
        count += 1

# Crear múltiples iteradores a partir del mismo generador
gen1, gen2 = itertools.tee(contador_infinito())    #lazy
#for i,j in zip(gen1,gen2):
#  print(i,j)

# Limitar cada generador de forma independiente
limited_gen1 = itertools.islice(gen1, 5)
print(list(limited_gen1)) # [0, 1, 2, 3, 4]

limited_gen2 = itertools.islice(gen2, 3)
print(list(limited_gen2)) # [0, 1, 2]
````

````python
# Usando send

def next_regla(tupla):
    len_tupla = len(tupla)
    i = 0
    while i < len_tupla:
        val = (yield tupla[i]) # El valor que se envía
        if val == 'reset':
            i = 0
        else:
            i += 1

# Crear una instancia
gen = next_regla((2, 3, 4, 0))

# Iterar normalmente
print(next(gen)) # 2
print(next(gen)) # 3

# Enviar el valor 'reset' para reiniciar el índice
print(gen.send("reset")) # 2
print(next(gen)) # 3
````

**Ventajas de los generadores:**
* No necesitan almacenar todos los datos en memoria
* Ideales para datasets grandes o infinitos
* Construyen pipelines eficientes
* Evitan carga RAM
* Integran mejor con streams, archivos, sockets, APIs
* Funcionan como máquinas de estados ligeras

````python
# send(), throw() y close()

def acumulador():
    total = 0
    while True:
        x = yield total
        total += x


gen = acumulador()
print(next(gen))       # inicia → 0
print(gen.send(10))    # envía 10 → 10
print(gen.send(5))     # envía 5  → 15
print(gen.send(3))     # envía 3  → 18
````

````python
#close()

def contador():
    try:
        n = 1
        while True:
            yield n
            n += 1
    except GeneratorExit:
        print("→ Generador cerrado limpiamente.")

gen = contador()

print(next(gen))  # 1
print(next(gen))  # 2

gen.close()  # Se lanza GeneratorExit dentro del generador
````

````python
# throw()

def procesar():
    try:
        while True:
            try:
                dato = yield
                print("Procesando:", dato)
            except ValueError:
                print("Se recibió ValueError dentro del generador.")
    finally:
        print("Finalizando generador.")

gen = procesar()
next(gen)                # Inicia el generador

gen.send("A")            # Procesando: A
gen.throw(ValueError)    # Inyecta excepción dentro del yield
gen.send("B")            # Procesando: B
gen.close()
````

````python
def ejemplo():
    try:
        while True:
            try:
                x = yield
                print("Recibido:", x)
            except RuntimeError:
                print("Excepción RuntimeError manejada dentro del generador.")
    except GeneratorExit:
        print("Generador cerrado via close().")
    finally:
        print("Limpieza final.")

gen = ejemplo()
next(gen)

gen.send(10)
gen.throw(RuntimeError)   # Inyectamos excepción manejada
gen.send(20)
gen.close()               # Cerramos generador
````

````python
from decimal import Decimal

def normalizar(factor):
    factor = Decimal(factor)
    return lambda x: Decimal(x) / factor

def centrar(media):
    media = Decimal(media)
    return lambda x: Decimal(x) - media

pipeline = [
    normalizar("100"),
    centrar("0.5"),
    abs
]

def aplicar_pipeline(funciones, valor):
    valor = Decimal(valor)
    for f in funciones:
        # abs() no acepta Decimal directo con lambda, pero sí funciona así
        valor = f(valor)
    return valor

resultado = aplicar_pipeline(pipeline, "40")
print(resultado)
````

## **Programación Orientada a Objetos**
Es un paradigma de programación que permite diseñar software a través de la interacción de datos u objetos.
Conceptos importantes en este paradigma son:

*   Clases
*   Objetos
*   Herencia
*   Polimorfismo
*   Métodos
*   Atributos









## **Objeto**
Es cualquier cosa tangible o intangible de la cual se pueden observar propiedades y comportamientos.

*Por ejemplo*: Un telefono es una cosa tangible que tiene como propiedades: **tamaño, modelo, marca**, entre otras. Además sus comportamientos o funcionalidades pueden ser: **llamar a un contacto, enviar mensaje, tomar foto**, entre otros.

## **Clase**
A los objetos los podemos organizar en colecciones o agrupamientos de objetos. Para realizar dicha organización se toman en cuenta características comunes o comportamientos comunes. Por tanto, una clase es una colección de datos u objetos que comparten características y comportamientos comunes.

Entonces podemos decir que un **objeto** pertenece o es parte de alguna **clase**.

*Ejemplo*: Las mascotas es una clase que agrupa animales que por sus características y comportamientos les asigmanos esa categoría.

![Imagen](https://i.ytimg.com/vi/KBTgL4jwY0o/hqdefault.jpg)

## **Atributos**

Una clase está compuesta de miembros de clase conocidos como atributos y métodos.

Los atributos definen las características que nos interesa representar en un objeto, eliminando aquellas que no nos interesan de dicho objeto. A este proceso se le conoce como *abstracción*.

Los atributos representan las características de los objetos en una clase a través de tipos de datos primitivos u de otros objetos.

## Herencia
La herencia es un concepto en programación orientada a objetos empleado para referirse a creación de una nueva clase a partir de una o más clases existentes. La nueva clase adquiere por "herencia" todos los atributos y comportamientos de la clase de la cual hereda. La clase que otorga la herencia se le conoce como ***clase padre***, ***clase base***, o ***super clase***. La clase que recibe la herencia se le conoce como ***clase hija***, ***clase derivada,*** ***sub clase*** o ***clase extendida***. La forma de indicar la herencia en código es la siguiente:


```
# class ClaseDerivada (ClaseBase):
```



Para ilustrar la estructura de la herencia de clases en Python veamos el siguiente código:

````python
class ClaseBase:
  data="125"

class ClaseExtendida(ClaseBase):
  pass

objetoClaseExtendida=ClaseExtendida()

print(objetoClaseExtendida.data) #observe como el objeto de la clase hija heredó el atributo 'data'
````

Ahora un ejemplo más completo. Vamos a crear nuestras clases bases. En este caso creamos la clase **Fruta** y la clase **Verdura**.

````python
class Fruta:
    def __init__(self,color):
        self.color=color

    def show(self):
        return f"Soy una fruta de color {self.color}"

class Verdura:
    def __init__(self,color):
        self.color=color

    def show(self):
        return f"Soy verdura de color {self.color}"
````

En el siguiente fragmento de código mostramos como aplicar herencia. Crearemos una clase **Manzana** que heredará los miembros(atributos y métodos) de la clase **Fruta**. Como el atributo color pertenece a la clase padre, entonces lo enviamos desde el constructor de Manzana hacia el constructor de Fruta.

````python
class Manzana(Fruta):
    def __init__(self,color):
        self.familia=""
        Fruta.__init__(self,color)
````

En Python es posible implementar herencia múltiple. A continuación un ejemplo con la clase Aguacate que algunos la consideran Fruta y otros Verdura. Con herencia múltiple se crea de la siguiente forma:

````python
class Aguacate(Fruta,Verdura):
    def __init__(self,color):
        Fruta.__init__(self,color)
        Verdura.__init__(self,color)
        self.familia=""

    def show(self):
        return Fruta.show(self)+", o bien,"+Verdura.show(self)

````

Ahora podemos crear objetos con las clases que hemos definido. Este proceso es igual que con las clases simples sin herencia.

````python
golden=Manzana("Verde")
print(golden.show())
golden.familia="Golden"
print(golden.familia)

hass=Aguacate("Verde")
print(hass.show())
````

Vamos a crear un **menu** con herencia....
Primero creamos la clase base:

````python
class Menu:
    def __init__(self,opciones,titulo='Menu de opciones'):
        self.opciones=opciones
        self.titulo=titulo

    def deplegarMenu(self):
        print("-"*len(self.titulo))
        print(self.titulo)
        print("-" * len(self.titulo))
        for i in range(len(self.opciones)):
            print("[{}] {}".format(i,self.opciones[i]))

    def leerOpcion(self):
        print("Elegir una opcion: ")
        while True:
            try:
                opcion = int(input())
                if opcion < len(self.opciones) and opcion >= 0:
                    return opcion
                else:
                    print("Opcion fuera del rango permitido")
            except:
                print("Opcion debe ser un entero")
````

Ahora vamos a crear una clase extendida heredando de la clase base **menu**

````python
class MenuHorizontal(Menu):
    def deplegarMenu(self):
        cars=[len(x)+6 for x in self.opciones]
        l=sum(cars)
        print("-"*l)
        cadena="{:^"+str(l)+"}"
        print(cadena.format(self.titulo))
        print("-" * l)
        cad=""
        for i in range(len(self.opciones)):
            cad+="[{}] {}  ".format(i,self.opciones[i])
        print(cad)
````

Finalmente podremos crear objetos de ambas clases.

````python
#Objeto de la clase MenuHorizontal
menuhorizontal=MenuHorizontal(['Archivo','Leer','Escribir','Salir'],'Seleccione una opcion')
menuhorizontal.deplegarMenu()
print(menuhorizontal.leerOpcion())
````
