## 1. Métricas centrales y de dispersión

### 1.1. Media

```python
def validar_datos(datos):
    if len(datos) == 0:
        raise ValueError("La lista no puede estar vacía")


def media(datos):
    validar_datos(datos)
    total = sum(datos)
    cantidad = len(datos)
    return total / cantidad


print(f"Media: {media(calificaciones):.4f}")
```

### 1.2. Mediana

```python
def mediana(datos):
    validar_datos(datos)
    ordenados = sorted(datos)
    cantidad = len(ordenados)
    centro = cantidad // 2

    if cantidad % 2 == 0:
        primero = ordenados[centro - 1]
        segundo = ordenados[centro]
        return (primero + segundo) / 2
    else:
        return ordenados[centro]


print("Calificaciones ordenadas:", sorted(calificaciones))
print("Mediana:", mediana(calificaciones))
```

### 1.3. Moda

```python
def moda(datos):
    validar_datos(datos)
    conteos = {}
    for valor in sorted(set(datos)):
        conteos[valor] = datos.count(valor)

    mayor_frecuencia = max(conteos.values())
    modas = []
    for valor, cantidad in conteos.items():
        if cantidad == mayor_frecuencia:
            modas.append(valor)

    return modas


print("Moda:", moda(calificaciones))
```

### 1.4. Rango

```python
def rango(datos):
    validar_datos(datos)
    mayor = max(datos)
    menor = min(datos)
    return mayor - menor


print("Rango:", rango(calificaciones))
```

### 1.5. Varianza poblacional

```python
def varianza_poblacional(datos):
    promedio = media(datos)
    suma_cuadrados = 0

    for valor in datos:
        diferencia = valor - promedio
        suma_cuadrados += diferencia ** 2

    return suma_cuadrados / len(datos)


print(f"Varianza poblacional: {varianza_poblacional(calificaciones):.4f}")
```

## 2. Distribución de frecuencias de calificaciones

### Diccionario automático mediante funciones de orden superior

```python
def frecuencias(datos):
    valores = sorted(set(datos))
    pares = map(lambda valor: (valor, datos.count(valor)), valores)
    return dict(pares)


distribucion = frecuencias(calificaciones)
print("Frecuencias:", distribucion)

for nota, cantidad in distribucion.items():
    porcentaje = cantidad / len(calificaciones) * 100
    print(f"Calificación {nota}: {cantidad} estudiantes ({porcentaje:.2f}%)")
```

## 3. Tokenización y análisis léxico

### 3.1. Conversión a minúsculas y eliminación de acentos y signos de puntuación

```python
def limpiar_texto(texto):
    texto = texto.lower()
    acentos = {"á": "a", "é": "e", "í": "i", "ó": "o", "ú": "u", "ü": "u"}

    for original, reemplazo in acentos.items():
        texto = texto.replace(original, reemplazo)

    texto_limpio = ""
    for caracter in texto:
        if caracter.isalnum() or caracter.isspace():
            texto_limpio += caracter
        else:
            texto_limpio += " "

    return " ".join(texto_limpio.split())


comentarios_limpios = list(map(limpiar_texto, comentarios))
print("Original:", comentarios[0])
print("Limpio:", comentarios_limpios[0])
```

### 3.2. Tokenización y conjunto de palabras únicas

```python
tokens_por_comentario = list(map(lambda texto: texto.split(), comentarios_limpios))

palabras = []
for tokens in tokens_por_comentario:
    for palabra in tokens:
        palabras.append(palabra)

palabras_unicas = set(palabras)

print("Palabras del primer comentario:", tokens_por_comentario[0])
print("Total de palabras:", len(palabras))
print("Cantidad de palabras distintas:", len(palabras_unicas))
print("Palabras únicas:", sorted(palabras_unicas))
```

### 3.3. Top 5 de palabras más frecuentes

```python
frecuencias_palabras = frecuencias(palabras)
ordenadas = sorted(frecuencias_palabras.items(), key=lambda par: (-par[1], par[0]))
top_5 = ordenadas[:5]

for palabra, cantidad in top_5:
    print(f"{palabra}: {cantidad}")
```

### 3.4. Vocabularios y porcentaje de palabras positivas y negativas

```python
positivas = {
    "buen", "bueno", "buenos", "excelente", "completo", "aprendi",
    "util", "genial", "recomendable", "recomiendo", "supero"
}
negativas = {"malo", "basico", "mejorable", "mejorar"}


def contar_palabras(tokens, vocabulario):
    encontradas = list(filter(lambda palabra: palabra in vocabulario, tokens))
    return len(encontradas)


def porcentaje_palabras(tokens, vocabulario):
    if len(tokens) == 0:
        return 0.0
    cantidad = contar_palabras(tokens, vocabulario)
    return cantidad / len(tokens) * 100


cantidad_positivas = contar_palabras(palabras, positivas)
cantidad_negativas = contar_palabras(palabras, negativas)
print("Apariciones positivas:", cantidad_positivas)
print("Apariciones negativas:", cantidad_negativas)
print(f"Porcentaje positivo: {porcentaje_palabras(palabras, positivas):.2f}%")
print(f"Porcentaje negativo: {porcentaje_palabras(palabras, negativas):.2f}%")
```

## 4. Clasificación automática de comentarios

### 4.1. Reglas de clasificación con operadores lógicos

```python
def clasificar_comentario(comentario, positivas, negativas):
    limpio = limpiar_texto(comentario)
    tokens = limpio.split()

    tiene_positivas = contar_palabras(tokens, positivas) > 0
    tiene_negativas = contar_palabras(tokens, negativas) > 0
    frase = " " + limpio + " "

    if " no lo recomiendo " in frase or " no recomiendo " in frase:
        return "Negativo"
    elif tiene_positivas and not tiene_negativas:
        return "Positivo"
    elif tiene_negativas and not tiene_positivas:
        return "Negativo"
    else:
        return "Neutro"


print(clasificar_comentario("Excelente contenido y profesor", positivas, negativas))
print(clasificar_comentario("No lo recomiendo", positivas, negativas))
print(clasificar_comentario("Curso bueno, aunque mejorable", positivas, negativas))
```

### 4.2. Agrupación en positivos, neutros y negativos

```python
grupos = {"Positivo": [], "Neutro": [], "Negativo": []}
etiquetas = []

for comentario in comentarios:
    categoria = clasificar_comentario(comentario, positivas, negativas)
    etiquetas.append(categoria)
    grupos[categoria].append(comentario)

for categoria, lista_comentarios in grupos.items():
    porcentaje = len(lista_comentarios) / len(comentarios) * 100
    print(f"\n{categoria}: {len(lista_comentarios)} ({porcentaje:.2f}%)")
    for comentario in lista_comentarios:
        print("-", comentario)
```
